# Week 5 Day 4: Categorical Encoding Techniques

Practice LabelEncoding (ordinal categorical data) and OneHotEncoding (nominal categorical data).

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# 1. Create Sample Categorical Data
data = {
    "Education": ["High School", "Bachelor", "Master", "PhD", "Bachelor"],
    "City": ["New York", "London", "Paris", "London", "New York"]
}
df = pd.DataFrame(data)

# 2. Label Encoding (for Ordinal Data: Education)
label_enc = LabelEncoder()
df["Education_LabelEncoded"] = label_enc.fit_transform(df["Education"])

# 3. One-Hot Encoding (for Nominal Data: City)
onehot_enc = OneHotEncoder(sparse_output=False)
city_encoded = onehot_enc.fit_transform(df[["City"]])
city_cols = onehot_enc.get_feature_names_out(["City"])
df_onehot = pd.DataFrame(city_encoded, columns=city_cols)

# 4. Concatenate Results
final_df = pd.concat([df, df_onehot], axis=1)

print("--- Categorical Encoding Output ---")
print(final_df.to_string(index=False))
```
