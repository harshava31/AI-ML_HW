# Week 5 Day 4: Data Preprocessing Revision

Repeat missing-value handling, encoding, scaling, and splitting in one unified workflow.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 1. Create Synthetic Dirty Dataset
raw_data = {
    "Age": [25, np.nan, 47, 38, 52, 29, 33, np.nan],
    "Income": [50000, 64000, np.nan, 82000, 110000, 48000, 71000, 95000],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "HR"],
    "Purchased": [0, 1, 1, 0, 1, 0, 1, 0]
}
df = pd.DataFrame(raw_data)

# 2. Handle Missing Values
num_imputer = SimpleImputer(strategy="median")
df[["Age", "Income"]] = num_imputer.fit_transform(df[["Age", "Income"]])

# 3. Categorical Encoding (One-Hot)
encoder = OneHotEncoder(sparse_output=False, drop="first")
dept_encoded = encoder.fit_transform(df[["Department"]])
dept_cols = encoder.get_feature_names_out(["Department"])
df_encoded = pd.DataFrame(dept_encoded, columns=dept_cols)

# 4. Combine & Feature Scaling
df_processed = pd.concat([df[["Age", "Income"]], df_encoded], axis=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_processed)
y = df["Purchased"]

# 5. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.25, random_state=42
)

print("--- Preprocessing Revision Completed ---")
print(f"X_train Shape: {X_train.shape} | X_test Shape: {X_test.shape}")
```
