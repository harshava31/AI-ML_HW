# Week 5 Day 2: Overfitting Concept Practice

Compare training and testing accuracy across Decision Trees and Random Forests to observe and mitigate overfitting.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# 1. Load Data
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.3, random_state=42, stratify=data.target
)

# 2. Evaluate Unconstrained Decision Tree vs. Random Forest
models = {
    "Unconstrained Decision Tree": DecisionTreeClassifier(random_state=42),
    "Constrained Decision Tree (max_depth=3)": DecisionTreeClassifier(max_depth=3, random_state=42),
    "Random Forest (n_estimators=100)": RandomForestClassifier(n_estimators=100, random_state=42),
}

overfitting_results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    train_acc = accuracy_score(y_train, model.predict(X_train))
    test_acc = accuracy_score(y_test, model.predict(X_test))
    gap = train_acc - test_acc
    
    overfitting_results.append({
        "Model": name,
        "Train Acc": f"{train_acc:.4f}",
        "Test Acc": f"{test_acc:.4f}",
        "Generalization Gap": f"{gap:.4f}"
    })

print("--- Overfitting Analysis ---")
print(pd.DataFrame(overfitting_results).to_string(index=False))
```
