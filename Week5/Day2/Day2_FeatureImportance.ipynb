# Week 5 Day 2: Feature Importance Visualization

Extract and visualize the top feature importances from the trained Random Forest model.

```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Load Dataset & Fit Model
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# 2. Extract Top 10 Features
importance_df = pd.DataFrame({
    "Feature": data.feature_names,
    "Importance": rf.feature_importances_
}).sort_values(by="Importance", ascending=False).head(10)

# 3. Plot & Save
plt.figure(figsize=(10, 6))
sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    palette="magma"
)
plt.title("Top 10 Feature Importances (Random Forest)", fontweight="bold")
plt.xlabel("Mean Decrease in Impurity (Gini Importance)")
plt.tight_layout()

chart_path = os.path.join(output_dir, "Day2_RF_FeatureImportance.png")
plt.savefig(chart_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Feature importance chart saved at: {chart_path}")
```
