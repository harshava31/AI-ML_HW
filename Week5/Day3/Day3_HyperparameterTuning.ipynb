# Week 5 Day 3: Hyperparameter Tuning Introduction

Run a basic `GridSearchCV` example to tune hyperparameters for a Random Forest classifier.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, train_test_split

# 1. Load Dataset & Train-Test Split
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)

# 2. Define Parameter Grid
param_grid = {
    "n_estimators": [50, 100, 150],
    "max_depth": [3, 5, 10, None],
    "min_samples_split": [2, 5],
    "criterion": ["gini", "entropy"],
}

# 3. Setup GridSearchCV
rf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1,
    verbose=1,
)

# 4. Execute Grid Search
grid_search.fit(X_train, y_train)

print(f"\nBest Cross-Validation Score: {grid_search.best_score_ * 100:.2f}%")
print(f"Best Hyperparameters: {grid_search.best_params_}")
```
