# Week 5 Day 3: Cross Validation Basics

Use `cross_val_score` to evaluate model stability and reliability across multiple data folds.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score

# 1. Load Dataset
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target

# 2. Instantiate Estimator & Stratified K-Fold
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 3. Compute 5-Fold Cross-Validation Scores
cv_scores = cross_val_score(rf_model, X, y, cv=cv, scoring="accuracy")

# 4. Display Results
print("--- 5-Fold Cross-Validation Results ---")
for fold, score in enumerate(cv_scores, start=1):
    print(f"Fold {fold}: {score * 100:.2f}%")

print(f"\nMean CV Accuracy: {np.mean(cv_scores) * 100:.2f}%")
print(f"Standard Deviation: {np.std(cv_scores):.4f}")
```
