# Week 5 Day 3: Accuracy Comparison Sheet

Maintain a table benchmarking baseline vs. cross-validated vs. hyperparameter-tuned models along with their accuracy scores.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import GridSearchCV, cross_val_score, train_test_split
from sklearn.tree import DecisionTreeClassifier

# 1. Load Data
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)

# 2. Benchmark Models
# Model 1: Default Decision Tree
dt = DecisionTreeClassifier(random_state=42).fit(X_train, y_train)
dt_pred = dt.predict(X_test)

# Model 2: Default Random Forest
rf_default = RandomForestClassifier(random_state=42).fit(X_train, y_train)
rf_def_pred = rf_default.predict(X_test)

# Model 3: Tuned Random Forest (via Grid Search)
param_grid = {"n_estimators": [50, 100], "max_depth": [3, 5, 8]}
grid = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5)
grid.fit(X_train, y_train)
rf_tuned = grid.best_estimator_
rf_tuned_pred = rf_tuned.predict(X_test)

# 3. Build Summary Table
comparison_table = pd.DataFrame([
    {
        "Model": "Decision Tree (Default)",
        "Train Acc": accuracy_score(y_train, dt.predict(X_train)),
        "Test Acc": accuracy_score(y_test, dt_pred),
        "F1 Score": f1_score(y_test, dt_pred),
    },
    {
        "Model": "Random Forest (Default)",
        "Train Acc": accuracy_score(y_train, rf_default.predict(X_train)),
        "Test Acc": accuracy_score(y_test, rf_def_pred),
        "F1 Score": f1_score(y_test, rf_def_pred),
    },
    {
        "Model": "Random Forest (Tuned)",
        "Train Acc": accuracy_score(y_train, rf_tuned.predict(X_train)),
        "Test Acc": accuracy_score(y_test, rf_tuned_pred),
        "F1 Score": f1_score(y_test, rf_tuned_pred),
    },
])

print("--- Model Performance Benchmarking Sheet ---")
print(comparison_table.to_string(index=False))
```
