# Week 5 Day 3: Best Parameters Analysis

Inspect best-performing hyperparameters from Grid Search and evaluate why they improve generalization performance over baseline models.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import GridSearchCV, train_test_split

# 1. Prepare Data
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)

# 2. Run Grid Search
param_grid = {
    "n_estimators": [50, 100, 150],
    "max_depth": [3, 5, 10],
    "min_samples_split": [2, 5],
}

grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
)
grid_search.fit(X_train, y_train)

# 3. Evaluate Best Estimator on Test Set
best_model = grid_search.best_estimator_
y_pred_best = best_model.predict(X_test)

print("--- Optimal Model Hyperparameters ---")
for param, val in grid_search.best_params_.items():
    print(f"• {param}: {val}")

print(f"\nTest Accuracy (Best Model): {accuracy_score(y_test, y_pred_best) * 100:.2f}%\n")
print("--- Classification Report ---")
print(classification_report(y_test, y_pred_best, target_names=data.target_names))
```
