# Week 4 Day 1: Feature and Target Identification
Identify independent and dependent variables in a machine learning dataset.

```python
import pandas as pd

# Create a sample synthetic dataset
data = {
    "Age": [25, 30, 45, 35, 22, 50],
    "Salary": [50000, 64000, 110000, 80000, 48000, 125000],
    "Experience_Years": [2, 5, 15, 8, 1, 20],
    "Purchased": [0, 0, 1, 1, 0, 1],  # Target variable
}

df = pd.DataFrame(data)

# Identifying Independent (Features) and Dependent (Target) Variables
feature_cols = ["Age", "Salary", "Experience_Years"]
target_col = "Purchased"

X = df[feature_cols]  # Independent variables
y = df[target_col]  # Dependent variable

print("Independent Variables (Features - X):")
print(X)

print("\nDependent Variable (Target - y):")
print(y)
```
