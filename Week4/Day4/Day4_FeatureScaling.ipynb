# Week 4 Day 4: Feature Scaling Methods

Compare StandardScaler (Standardization) vs MinMaxScaler (Normalization).

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {
    "Age": [22, 45, 30, 52, 25],
    "EstimatedSalary": [18000, 120000, 65000, 95000, 32000],
}
df = pd.DataFrame(data)

std_scaler = StandardScaler()
df_std = pd.DataFrame(std_scaler.fit_transform(df), columns=["Age_Std", "Salary_Std"])

minmax_scaler = MinMaxScaler()
df_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(df), columns=["Age_MinMax", "Salary_MinMax"]
)

combined = pd.concat([df, df_std, df_minmax], axis=1)
print("--- Feature Scaling Comparison ---")
print(combined)
```
