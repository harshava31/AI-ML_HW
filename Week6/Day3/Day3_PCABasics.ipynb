# Week 6 Day 3: PCA Basics

Learn dimensionality reduction fundamentals and apply Principal Component Analysis (PCA) to a multi-feature dataset.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Load High-Dimensional Dataset (30 features)
data = load_breast_cancer()
X = data.data
y = data.target

# 2. Standardize Features (Crucial for PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply PCA (reduce to 2 principal components)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("--- PCA Transformation Summary ---")
print(f"Original Feature Shape : {X.shape}")
print(f"Reduced PCA Shape      : {X_pca.shape}")
print(f"Explained Variance Ratio: {pca.explained_variance_ratio_.round(4)}")
```
