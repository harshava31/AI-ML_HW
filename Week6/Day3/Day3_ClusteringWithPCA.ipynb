# Week 6 Day 3: Clustering with PCA-Transformed Data

Run KMeans clustering on PCA-reduced feature spaces and compare clustering efficiency and visualizations.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Scale Data & Reduce Dimensions with PCA
data = load_breast_cancer()
X_scaled = StandardScaler().fit_transform(data.data)
X_pca = PCA(n_components=2).fit_transform(X_scaled)

# 2. Apply KMeans Clustering on PCA-transformed Features
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_pca)

df_clusters = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
df_clusters["Cluster"] = cluster_labels

# 3. Visualize Clusters in PCA Space
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_clusters,
    x="PC1",
    y="PC2",
    hue="Cluster",
    palette="Set2",
    s=60,
    alpha=0.8,
)
plt.title("KMeans Clustering on PCA-Reduced Space", fontweight="bold")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Cluster")
plt.tight_layout()

chart_path = os.path.join(output_dir, "Day3_ClusteringWithPCA.png")
plt.savefig(chart_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Clustering with PCA plot saved at: {chart_path}")
```
