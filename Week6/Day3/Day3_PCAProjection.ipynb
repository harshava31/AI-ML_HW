# Week 6 Day 3: 2D PCA Projection Visualization

Transform high-dimensional data into two principal components and visualize the projection using a 2D scatter plot.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Load, Scale, and Project to 2D
data = load_breast_cancer()
X_scaled = StandardScaler().fit_transform(data.data)
X_pca = PCA(n_components=2).fit_transform(X_scaled)

df_pca = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
df_pca["Target"] = data.target

# 2. Plot 2D PCA Projection
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="Target",
    palette="coolwarm",
    alpha=0.8,
    s=60,
)
plt.title("2D PCA Projection of Breast Cancer Dataset", fontweight="bold")
plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")
plt.legend(title="Class", labels=["Malignant", "Benign"])
plt.tight_layout()

chart_path = os.path.join(output_dir, "Day3_PCA_Projection.png")
plt.savefig(chart_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"PCA projection chart saved successfully at: {chart_path}")
```
