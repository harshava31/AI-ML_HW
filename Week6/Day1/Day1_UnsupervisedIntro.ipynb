# Week 6 Day 1: Unsupervised Learning Introduction

Review the foundational concepts of unsupervised learning and practice grouping data points without target labels.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_blobs

# 1. Generate Unlabeled Synthetic Dataset
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

df_unlabeled = pd.DataFrame(X, columns=["Feature_1", "Feature_2"])

# 2. Inspect Dataset Properties
print("--- Unlabeled Dataset Preview ---")
print(df_unlabeled.head())
print(f"\nDataset Shape: {df_unlabeled.shape}")
print(f"Number of Dimensions: {df_unlabeled.ndim}")
```
