# Week 6 Day 1: KMeans Clustering Basics

Apply KMeans clustering to sample data and inspect cluster assignments and centroid locations.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Generate Sample Data
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.70, random_state=42)

# 2. Apply KMeans Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

# 3. Extract Cluster Labels and Centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# 4. Display Results Summary
print("--- KMeans Clustering Summary ---")
print(f"Cluster Assignments (First 10 samples):\n{labels[:10]}")
print(f"\nCluster Centroids Coordinates:\n{centroids}")
```
