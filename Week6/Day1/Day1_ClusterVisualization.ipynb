# Week 6 Day 1: Cluster Visualization

Create a scatter plot clearly displaying the data points grouped by clusters alongside their computed centroids for a two-feature dataset.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Ensure output directory exists
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Generate Data & Fit KMeans
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.65, random_state=42)
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)
centroids = kmeans.cluster_centers_

# 2. Plot Clusters and Centroids
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X[:, 0],
    y=X[:, 1],
    hue=y_kmeans,
    palette="viridis",
    s=50,
    alpha=0.8,
    legend="full",
)

# Plot Centroids
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c="red",
    s=200,
    marker="X",
    label="Centroids",
)

plt.title("KMeans Cluster Visualization", fontweight="bold")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.tight_layout()

# 3. Save Figure
chart_path = os.path.join(output_dir, "Day1_KMeans_Clusters.png")
plt.savefig(chart_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Cluster visualization saved successfully at: {chart_path}")
```
