# Week 6 Day 4: Distance-Based Outlier Check

Detect far-away points in a small numerical dataset using Euclidean distance logic from the data centroid.

```python
import numpy as np
import pandas as pd

# 1. Create a Small Numerical Dataset
X = np.array([
    [2.0, 2.0],
    [2.5, 2.2],
    [2.2, 2.8],
    [2.1, 2.3],
    [10.0, 12.0]  # Clear outlier point
])

df = pd.DataFrame(X, columns=["Feature_1", "Feature_2"])

# 2. Compute Centroid and Euclidean Distances
centroid = df.mean(axis=0).values
df["Euclidean_Distance"] = np.linalg.norm(df.values - centroid, axis=1)

# 3. Flag Outliers (e.g., Distance > 2 * Mean Distance)
mean_dist = df["Euclidean_Distance"].mean()
df["Is_Distant_Outlier"] = df["Euclidean_Distance"] > (2 * mean_dist)

print("--- Distance-Based Outlier Analysis ---")
print(df)
```
