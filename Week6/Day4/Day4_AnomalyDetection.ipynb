# Week 6 Day 4: Anomaly Detection Introduction

Use a basic Z-score or IQR statistical approach to detect and isolate unusual outlier values in a dataset.

```python
import numpy as np
import pandas as pd

# 1. Generate Synthetic Dataset with Outliers
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=200)
data_with_outliers = np.append(data, [105.5, -15.0, 115.0]) # Inject anomalies

df = pd.DataFrame(data_with_outliers, columns=["Value"])

# 2. Compute Z-Scores
mean_val = df["Value"].mean()
std_val = df["Value"].std()
df["Z_Score"] = (df["Value"] - mean_val) / std_val

# 3. Identify Anomalies (|Z-Score| > 3)
df["Is_Anomaly"] = df["Z_Score"].abs() > 3

print("--- Anomaly Detection Summary ---")
print(f"Total Anomalies Detected: {df['Is_Anomaly'].sum()}")
print(df[df["Is_Anomaly"]])
```
