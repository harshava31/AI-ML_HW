# Week 6 Day 2: Standardization Before Clustering

Compare KMeans clustering results and boundary formations before and after applying `StandardScaler` feature scaling.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Create Unscaled Dataset with Wide Disparities in Scale
data = {
    "Age": [22, 25, 47, 52, 29, 33, 41, 60],
    "Annual_Income_k$": [18, 22, 120, 135, 25, 30, 95, 140],
}
df = pd.DataFrame(data)

# 2. Cluster Without Scaling
kmeans_unscaled = KMeans(n_clusters=2, random_state=42, n_init=10)
df["Cluster_Unscaled"] = kmeans_unscaled.fit_predict(df[["Age", "Annual_Income_k$"]])

# 3. Cluster With Standardization (`StandardScaler`)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[["Age", "Annual_Income_k$"]])

kmeans_scaled = KMeans(n_clusters=2, random_state=42, n_init=10)
df["Cluster_Scaled"] = kmeans_scaled.fit_predict(X_scaled)

print("--- Feature Scaling Impact Comparison ---")
print(df)
```
