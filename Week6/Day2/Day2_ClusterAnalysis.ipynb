# Week 6 Day 2: Cluster Label Analysis

Add predicted cluster labels back to the dataset and analyze statistical characteristics of each cluster group.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Generate Dataset & Fit KMeans (K=3)
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)
df = pd.DataFrame(X, columns=["Feature_A", "Feature_B"])

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X)

# 2. Analyze Cluster Groups (Mean values per cluster)
cluster_summary = df.groupby("Cluster").mean()

print("--- Cluster Group Summary Statistics ---")
print(cluster_summary)

print("\n--- Value Counts per Cluster ---")
print(df["Cluster"].value_counts())
```
