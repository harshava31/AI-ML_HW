# Week 6 Day 2: Real Dataset Customer Clustering

Apply KMeans clustering to a customer behavioral dataset to segment distinct spending groups.

```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Simulate a Customer Dataset (or load via pd.read_csv)
data = {
    "Customer_ID": range(1, 201),
    "Age": [20, 35, 40, 23, 55, 48, 22, 33, 45, 60] * 20,
    "Annual_Income_k$": [15, 60, 85, 18, 110, 95, 20, 55, 90, 125] * 20,
    "Spending_Score_1_100": [80, 50, 45, 85, 20, 30, 78, 52, 40, 15] * 20,
}
df = pd.DataFrame(data)

# 2. Select Features & Scale
features = ["Annual_Income_k$", "Spending_Score_1_100"]
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply KMeans (K=5 segments)
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["Segment"] = kmeans.fit_predict(X_scaled)

# 4. Visualize Customer Segments
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="Annual_Income_k$",
    y="Spending_Score_1_100",
    hue="Segment",
    palette="Set1",
    s=70,
    alpha=0.9,
)
plt.title("Customer Segmentation via KMeans Clustering", fontweight="bold")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.legend(title="Segment")
plt.tight_layout()

chart_path = os.path.join(output_dir, "Day2_CustomerSegments.png")
plt.savefig(chart_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Customer segment plot saved successfully at: {chart_path}")
```
