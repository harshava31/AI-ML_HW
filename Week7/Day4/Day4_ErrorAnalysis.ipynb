# Week 7 Day 4: Error Analysis

Inspect wrong predictions and suggest possible reasons for the errors.

```python
import numpy as np
import tensorflow as tf

mnist = tf.keras.datasets.mnist
(_, _), (X_test, y_test) = mnist.load_data()
X_test_scaled = X_test / 255.0

# Using reloaded or trained model to predict
model = tf.keras.models.load_model("saved_model.keras")
y_pred_classes = np.argmax(model.predict(X_test_scaled), axis=1)

incorrect_indices = np.where(y_pred_classes != y_test)[0]
print(f"Total misclassified samples: {len(incorrect_indices)}")

print("\n--- First 5 Error Samples ---")
for idx in incorrect_indices[:5]:
    print(f"Index {idx}: True = {y_test[idx]} | Predicted = {y_pred_classes[idx]}")
```
