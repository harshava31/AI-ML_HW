# Week 7 Day 1: Perceptron and MLP Basics

Build a basic Multi-Layer Perceptron model on a small binary classification dataset.

```python
from sklearn.datasets import load_iris
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.25, random_state=42, stratify=iris.target
)

mlp_clf = MLPClassifier(hidden_layer_sizes=(10, 5), max_iter=500, random_state=42)
mlp_clf.fit(X_train, y_train)

y_pred = mlp_clf.predict(X_test)
print("--- MLP Classification Report ---")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
```
