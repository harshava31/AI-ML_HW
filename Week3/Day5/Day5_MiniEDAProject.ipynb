# Week 3 Day 5: Mini EDA Project
Perform exploratory data analysis on a sample dataset and include charts and findings.

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# 1. Create and save sample e-commerce dataset
np.random.seed(42)
data = {
    "Order_ID": range(1001, 1201),
    "Category": np.random.choice(["Electronics", "Clothing", "Home", "Books"], size=200),
    "Sales": np.random.exponential(scale=100, size=200) + 10,
    "Discount": np.random.choice([0.0, 0.1, 0.2, 0.3], size=200),
    "Customer_Rating": np.random.uniform(3.0, 5.0, size=200)
}
df = pd.DataFrame(data)

# 2. Data Clean-up & Calculation
df["Net_Sales"] = df["Sales"] * (1 - df["Discount"])

# 3. Exploratory Charts
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Plot 1: Category Sales
sns.barplot(data=df, x="Category", y="Net_Sales", palette="viridis", ax=axes[0], errorbar=None)
axes[0].set_title("Average Net Sales by Category", fontweight="bold")

# Plot 2: Correlation Heatmap
sns.heatmap(df[["Sales", "Discount", "Customer_Rating", "Net_Sales"]].corr(), annot=True, cmap="coolwarm", ax=axes[1])
axes[1].set_title("Feature Correlations", fontweight="bold")

plt.tight_layout()
plt.savefig(os.path.join(output_dir, "Day5_MiniEDA_Project.png"), dpi=300, bbox_inches="tight")
plt.show()

print("--- EDA Project Executed Successfully ---")
```
