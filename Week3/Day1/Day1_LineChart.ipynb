## Task 1: Matplotlib Basics (Line Chart)
Creating a simple line chart with title, axis labels, markers, and readable formatting.

```python
# Sample data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperatures = [22, 24, 19, 23, 25, 28, 26]

# Create figure and plot line
plt.figure(figsize=(8, 4.5))
plt.plot(
    days,
    temperatures,
    color="royalblue",
    marker="o",
    linestyle="-",
    linewidth=2,
    markersize=6,
)

# Adding labels and title
plt.title("Weekly Temperature Trend", fontsize=14, fontweight="bold")
plt.xlabel("Day of the Week", fontsize=12)
plt.ylabel("Temperature (°C)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
