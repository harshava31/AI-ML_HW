# Week 3 Day 1: Histogram Plot
Create a histogram using random or sample numeric data and experiment with bins.

```python
import matplotlib.pyplot as plt
import numpy as np

# Set seed for reproducible random dataset
np.random.seed(42)

# Generate 500 normal distribution data points
exam_scores = np.random.normal(loc=75, scale=10, size=500)

# Create figure and histogram with bin experimentation
plt.figure(figsize=(8.5, 4.5))
plt.hist(
    exam_scores,
    bins=15,
    color="darkorange",
    edgecolor="black",
    alpha=0.75,
)

# Customize formatting
plt.title("Distribution of Student Exam Scores", fontsize=14, fontweight="bold")
plt.xlabel("Score Range", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)
plt.grid(axis="y", linestyle=":", alpha=0.7)

plt.tight_layout()
plt.show()
```
