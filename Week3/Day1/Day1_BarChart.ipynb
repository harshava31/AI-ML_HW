# Week 3 Day 1: Bar Chart Program
Build a bar chart using categories and values.

```python
import matplotlib.pyplot as plt

# Sample categories and numerical values
products = ["Laptops", "Smartphones", "Tablets", "Headphones", "Monitors"]
sales_units = [120, 250, 90, 310, 150]

# Create figure and vertical bar chart
plt.figure(figsize=(8.5, 4.5))
bars = plt.bar(products, sales_units, color="teal", edgecolor="black", width=0.6)

# Adding values on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        yval + 5,
        f"{yval}",
        ha="center",
        va="bottom",
        fontsize=10,
    )

# Formatting title and axis labels
plt.title("Product Sales Units Summary", fontsize=14, fontweight="bold")
plt.xlabel("Product Category", fontsize=12)
plt.ylabel("Units Sold", fontsize=12)
plt.ylim(0, max(sales_units) + 50)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
