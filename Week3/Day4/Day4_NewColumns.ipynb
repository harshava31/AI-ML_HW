# Week 2 Day 4: New Column Creation
Create new columns based on calculations from existing columns, such as total marks or percentages.

```python
import pandas as pd

# 1. Create dataset with student marks
data = {
    "Student_Name": ["Arun", "Bhavna", "Chetan", "Divya"],
    "Math": [85, 92, 78, 65],
    "Science": [90, 88, 82, 70],
    "English": [88, 95, 75, 80],
}

df = pd.DataFrame(data)
max_marks_per_subject = 100
total_possible = max_marks_per_subject * 3

# 2. Create new column for Total Marks
df["Total_Marks"] = df["Math"] + df["Science"] + df["English"]

# 3. Create new column for Percentage
df["Percentage"] = (df["Total_Marks"] / total_possible) * 100

# 4. Conditional column creation based on percentage
df["Status"] = df["Percentage"].apply(
    lambda x: "Pass" if x >= 75 else "Needs Improvement"
)

print("--- Updated DataFrame with Computed Columns ---")
print(df)
```
