# Week 3 Day 4: EDA Summary
Write five key observations about a dataset and support them with charts.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

df = sns.load_dataset("titanic")

# Visual support for observations
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

sns.barplot(data=df, x="sex", y="survived", palette="muted", ax=axes[0], errorbar=None)
axes[0].set_title("Survival Rate by Sex", fontweight="bold")

sns.boxplot(data=df, x="pclass", y="age", palette="Set2", ax=axes[1])
axes[1].set_title("Age Distribution across Passenger Classes", fontweight="bold")

plt.tight_layout()
plt.savefig(os.path.join(output_dir, "Day4_EDASummaryPlots.png"), dpi=300, bbox_inches="tight")
plt.show()
```

## Key Observations Summary

1. **Sex and Survival Rate**: Female passengers had a significantly higher survival rate (>70%) compared to male passengers (<20%).
2. **Passenger Class Impact**: 1st Class passengers had a higher average age and higher overall survival rates than 3rd Class passengers.
3. **Age Distribution**: 3rd Class passengers were predominantly younger (median age around 24), whereas 1st Class passengers were older (median age around 38).
4. **Missing Values**: The `deck` column contains over 75% missing data, requiring dropping or substantial imputation.
5. **Fare Skewness**: Fares are strongly right-skewed with a few high outliers corresponding to 1st Class suites.
