# Week 2 Day 4: Missing Values Basics
Create missing values in a dataset and practice identifying, filling, and dropping them.

```python
import numpy as np
import pandas as pd

# 1. Create a dataset containing missing values (NaN)
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Math_Score": [85, np.nan, 90, 78, np.nan],
    "Science_Score": [92, 88, np.nan, 75, 82],
    "Attendance": [95, 80, 85, np.nan, 90],
}

df = pd.DataFrame(data)
print("--- Original DataFrame with Missing Values ---")
print(df)

# 2. Identify missing values
print("\n--- Missing Value Counts Per Column ---")
print(df.isnull().sum())

# 3. Fill missing values (Imputation)
# Fill numeric scores with column mean
df_filled = df.copy()
df_filled["Math_Score"] = df_filled["Math_Score"].fillna(
    df_filled["Math_Score"].mean()
)
df_filled["Science_Score"] = df_filled["Science_Score"].fillna(
    df_filled["Science_Score"].mean()
)
print("\n--- DataFrame After Filling Missing Scores with Mean ---")
print(df_filled)

# 4. Drop rows with any remaining missing values
df_dropped = df.dropna()
print("\n--- DataFrame After Dropping Any Row with NaN ---")
print(df_dropped)
```
