# Week 2 Day 4: Sorting and Renaming
Rename columns and sort records based on values in one or more columns.

```python
import pandas as pd

# 1. Create sample DataFrame
data = {
    "emp_id": [1, 2, 3, 4, 5],
    "fname": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "dept": ["IT", "HR", "IT", "Finance", "HR"],
    "sal": [85000, 62000, 92000, 75000, 68000],
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)

# 2. Rename columns to improve readability
df_renamed = df.rename(
    columns={
        "emp_id": "Employee_ID",
        "fname": "First_Name",
        "dept": "Department",
        "sal": "Salary",
    }
)
print("\n--- Renamed DataFrame ---")
print(df_renamed)

# 3. Sort by single column (Salary descending)
sorted_salary = df_renamed.sort_values(by="Salary", ascending=False)
print("\n--- Sorted by Salary (Descending) ---")
print(sorted_salary)

# 4. Sort by multiple columns (Department ascending, Salary descending)
sorted_multi = df_renamed.sort_values(
    by=["Department", "Salary"], ascending=[True, False]
)
print("\n--- Sorted by Department (Asc) & Salary (Desc) ---")
print(sorted_multi)
```
