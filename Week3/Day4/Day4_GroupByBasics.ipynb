# Week 2 Day 4: Grouping Basics
Use groupby to calculate counts and averages by category.

```python
import pandas as pd

# 1. Create sample dataset
data = {
    "Store": [
        "North",
        "North",
        "South",
        "South",
        "East",
        "East",
        "North",
        "East",
    ],
    "Category": [
        "Electronics",
        "Furniture",
        "Electronics",
        "Furniture",
        "Electronics",
        "Furniture",
        "Electronics",
        "Furniture",
    ],
    "Sales": [1200, 800, 1500, 400, 950, 1100, 1300, 700],
    "Quantity": [10, 5, 12, 3, 8, 9, 11, 6],
}

df = pd.DataFrame(data)
print("--- Original Sales Dataset ---")
print(df)

# 2. Groupby single column: Average sales by Store
avg_sales_by_store = df.groupby("Store")["Sales"].mean().reset_index()
print("\n--- Average Sales by Store ---")
print(avg_sales_by_store)

# 3. Groupby with multiple aggregations (count and mean)
grouped_summary = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Average_Sales=("Sales", "mean"),
        Transaction_Count=("Sales", "count"),
    )
    .reset_index()
)

print("\n--- Aggregated Summary by Category ---")
print(grouped_summary)
```
