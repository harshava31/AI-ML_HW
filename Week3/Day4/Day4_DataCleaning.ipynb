# Week 3 Day 4: Data Cleaning Practice
Identify and remove duplicate rows and standardize column formats.

```python
import pandas as pd

# Create dataset with duplicates and inconsistent formatting
raw_data = {
    "Customer_ID": [101, 102, 102, 103, 104, 105],
    " Name ": [" Alice ", "BOB", "bob", "Charlie", "David ", "Eva"],
    "City": ["New York", "Chicago", "Chicago", "NEW YORK", "Houston", "houston"],
    "Signup_Date": ["2023-01-15", "2023-02-20", "2023-02-20", "2023-03-05", "2023-04-12", "2023-05-01"],
}

df = pd.DataFrame(raw_data)
print("--- Raw DataFrame ---")
print(df)

# 1. Standardize column names (strip whitespace and lower/title case)
df.columns = df.columns.str.strip().str.title()

# 2. Clean string formats in columns
df["Name"] = df["Name"].str.strip().str.title()
df["City"] = df["City"].str.strip().str.title()

# 3. Identify and drop duplicates
print("\n--- Duplicate Rows Count ---")
print(df.duplicated().sum())

df_cleaned = df.drop_duplicates().reset_index(drop=True)

print("\n--- Cleaned DataFrame ---")
print(df_cleaned)
```
