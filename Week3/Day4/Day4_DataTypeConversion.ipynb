# Week 3 Day 4: Data Type Conversion
Convert strings to numeric values, parse dates, and adjust category types.

```python
import pandas as pd

# Raw data with string types for numbers and dates
data = {
    "Transaction_ID": ["1001", "1002", "1003", "1004"],
    "Price": ["$29.99", "$49.50", "$15.00", "$99.99"],
    "Date_Str": ["2024-01-15", "2024-02-01", "2024-02-18", "2024-03-05"],
    "Priority": ["Low", "High", "Medium", "High"]
}

df = pd.DataFrame(data)
print("--- Initial Data Types ---")
print(df.dtypes)

# 1. Convert Price string to float
df["Price"] = df["Price"].str.replace("$", "", regex=False).astype(float)

# 2. Parse Date strings to datetime objects
df["Date"] = pd.to_datetime(df["Date_Str"])
df = df.drop(columns=["Date_Str"])

# 3. Convert Priority to Categorical type with ordering
priority_order = ["Low", "Medium", "High"]
df["Priority"] = pd.Categorical(df["Priority"], categories=priority_order, ordered=True)

print("\n--- Converted Data Types ---")
print(df.dtypes)
print("\n--- Processed DataFrame ---")
print(df)
```
