# Week 2 Day 3: Pandas Series Practice
Create pandas Series objects and practice indexing, filtering, and value access.

```python
import pandas as pd

# 1. Create a Pandas Series with custom index
data = [100, 250, 400, 150, 300]
labels = ["Mon", "Tue", "Wed", "Thu", "Fri"]
series = pd.Series(data, index=labels, name="Daily_Sales")

print("--- Original Series ---")
print(series)

# 2. Value Access and Indexing
print("\n--- Indexing Examples ---")
print("Sales on Wednesday:", series["Wed"])
print("First three days using positional slice:\n", series.iloc[:3])

# 3. Filtering with Boolean Conditions
filtered_series = series[series > 200]
print("\n--- Filtered Days (Sales > 200) ---")
print(filtered_series)
```
