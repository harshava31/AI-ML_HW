# Week 3 Day 3: Correlation Heatmap
Create a correlation matrix and visualize it with a heatmap.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# Load dataset
df = sns.load_dataset("iris")

# Compute Correlation Matrix for numerical columns
corr_matrix = df.drop(columns=["species"]).corr()

print("--- Correlation Matrix ---")
print(corr_matrix)

# Visualize Heatmap
fig, ax = plt.subplots(figsize=(7, 5))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": 0.8},
    ax=ax,
)

ax.set_title("Feature Correlation Heatmap", fontweight="bold")
plt.tight_layout()
plt.savefig(
    os.path.join(output_dir, "Day3_CorrelationHeatmap.png"),
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```
