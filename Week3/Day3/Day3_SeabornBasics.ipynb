# Week 3 Day 3: Seaborn Basics
Practice count plots, box plots, and pair plots using a sample dataset.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# Load sample dataset
df = sns.load_dataset("tips")

# 1. Count Plot
fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(data=df, x="day", hue="sex", palette="Set2", ax=ax)
ax.set_title("Customer Count by Day and Sex", fontweight="bold")
plt.tight_layout()
plt.savefig(
    os.path.join(output_dir, "Day3_CountPlot.png"), dpi=300, bbox_inches="tight"
)
plt.show()

# 2. Box Plot
fig, ax = plt.subplots(figsize=(6, 4))
sns.boxplot(data=df, x="day", y="total_bill", palette="Pastel1", ax=ax)
ax.set_title("Total Bill Distribution by Day", fontweight="bold")
plt.tight_layout()
plt.savefig(
    os.path.join(output_dir, "Day3_BoxPlot.png"), dpi=300, bbox_inches="tight"
)
plt.show()

# 3. Pair Plot
pair_plot = sns.pairplot(df, hue="species" if "species" in df else "sex", palette="husl")
pair_plot.fig.suptitle("Pair Plot of Numerical Features", y=1.02, fontweight="bold")
pair_plot.savefig(os.path.join(output_dir, "Day3_PairPlot.png"), dpi=300, bbox_inches="tight")
plt.show()
```
