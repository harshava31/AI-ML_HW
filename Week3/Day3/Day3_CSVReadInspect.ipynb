# Week 2 Day 3: CSV Read and Inspect
Create or load a CSV file and inspect it using head, tail, info, and describe.

```python
import os
import pandas as pd

# 1. Generate a sample CSV file locally
sample_data = """ProductID,Category,Price,UnitsSold
P001,Electronics,299.99,45
P002,Clothing,19.99,120
P003,Electronics,899.00,15
P004,Home,45.50,80
P005,Clothing,34.00,65
P006,Home,12.99,200
"""

csv_path = "sample_data.csv"
with open(csv_path, "w") as f:
    f.write(sample_data)

# 2. Read CSV into Pandas DataFrame
df = pd.read_csv(csv_path)

# 3. Inspection Operations
print("--- First 3 Rows (head) ---")
print(df.head(3))

print("\n--- Last 2 Rows (tail) ---")
print(df.tail(2))

print("\n--- Summary Information (info) ---")
df.info()

print("\n--- Descriptive Statistics (describe) ---")
print(df.describe())
```
