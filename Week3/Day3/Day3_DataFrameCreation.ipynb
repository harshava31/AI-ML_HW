# Week 2 Day 3: DataFrame Creation
Create DataFrames from dictionaries and lists and inspect rows and columns.

```python
import pandas as pd

# 1. Create DataFrame from a dictionary of lists
data_dict = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Department": ["HR", "IT", "Finance", "IT", "Marketing"],
    "Salary": [65000, 85000, 75000, 90000, 70000],
}

df = pd.DataFrame(data_dict)

print("--- DataFrame ---")
print(df)

# 2. Inspect Rows and Columns
print("\n--- Structural Overview ---")
print("Shape (rows, columns):", df.shape)
print("Columns list:", list(df.columns))
print("Index range:", df.index)
print("Data types:\n", df.dtypes)
```
