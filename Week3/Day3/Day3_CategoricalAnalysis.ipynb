# Week 3 Day 3: Categorical Analysis
Analyze categorical features by counts or averages and present the results visually.

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# Load dataset
df = sns.load_dataset("titanic")

# Aggregate average survival rate by class and sex
grouped = (
    df.groupby(["class", "sex"], observed=False)["survived"].mean().reset_index()
)
print("--- Survival Rate by Class and Sex ---")
print(grouped)

# Visualizing Average Survival Rate
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.barplot(
    data=df,
    x="class",
    y="survived",
    hue="sex",
    palette="muted",
    errorbar=None,
    ax=ax,
)

ax.set_title("Survival Rate by Passenger Class & Gender", fontweight="bold")
ax.set_ylabel("Average Survival Rate")
ax.set_xlabel("Passenger Class")

plt.tight_layout()
plt.savefig(
    os.path.join(output_dir, "Day3_CategoricalAnalysis.png"),
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```
