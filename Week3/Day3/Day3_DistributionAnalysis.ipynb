# Week 3 Day 3: Distribution Analysis
Analyze the distribution of numerical columns and interpret mean/median behavior.

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="ticks")
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

# Generate skewed sample data
np.random.seed(42)
income_data = np.random.exponential(scale=50000, size=500) + 20000
df = pd.DataFrame({"Income": income_data})

# Calculate Metrics
mean_val = df["Income"].mean()
median_val = df["Income"].median()

print(f"Mean Income: ${mean_val:,.2f}")
print(f"Median Income: ${median_val:,.2f}")

# Plot Histogram with KDE
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.histplot(df["Income"], kde=True, color="teal", ax=ax)

# Annotate Mean and Median
ax.axvline(
    mean_val, color="red", linestyle="--", linewidth=2, label=f"Mean: ${mean_val:,.0f}"
)
ax.axvline(
    median_val,
    color="orange",
    linestyle="-",
    linewidth=2,
    label=f"Median: ${median_val:,.0f}",
)

ax.set_title(
    "Income Distribution & Skewness Analysis (Mean vs Median)",
    fontweight="bold",
)
ax.set_xlabel("Income ($)")
ax.legend()

plt.tight_layout()
plt.savefig(
    os.path.join(output_dir, "Day3_DistributionAnalysis.png"),
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```
