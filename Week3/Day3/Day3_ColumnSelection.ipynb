# Week 2 Day 3: Column Selection
Select single and multiple columns and filter rows with conditions.

```python
import pandas as pd

# Sample Dataset
data = {
    "Title": ["Python Basics", "Data Analysis", "Deep Learning", "Web Dev", "SQL Guide"],
    "Category": ["Programming", "Data Science", "AI/ML", "Web", "Data Science"],
    "Price": [29.99, 49.99, 89.99, 39.99, 19.99],
    "Rating": [4.5, 4.8, 4.9, 4.2, 4.6],
}

df = pd.DataFrame(data)

# 1. Single Column Selection (Series output)
titles = df["Title"]
print("--- Single Column (Title) ---")
print(titles.head())

# 2. Multiple Column Selection (DataFrame output)
subset = df[["Title", "Price", "Rating"]]
print("\n--- Multiple Columns Subset ---")
print(subset)

# 3. Filtering Rows with Boolean Logic
print("\n--- Filtering: High-rated books (Rating >= 4.7) ---")
high_rated = df[df["Rating"] >= 4.7]
print(high_rated)

print("\n--- Compound Filtering: Data Science category AND Price < $30 ---")
affordable_ds = df[(df["Category"] == "Data Science") & (df["Price"] < 30)]
print(affordable_ds)
```
