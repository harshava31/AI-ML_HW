# Week 3 Day 2: Plot Customization
Add legends, gridlines, markers, and figure-size changes to improve readability.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(9, 5))

# Plot multi-series with markers, colors, and line styles
plt.plot(
    x,
    y1,
    label="Sine Wave",
    color="darkblue",
    linestyle="-",
    linewidth=2.5,
    marker="o",
    markevery=10,
)
plt.plot(
    x,
    y2,
    label="Cosine Wave",
    color="darkorange",
    linestyle="--",
    linewidth=2.5,
    marker="s",
    markevery=10,
)

# Custom formatting
plt.title(
    "Customized Trigonometric Waveforms",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("X Value (Radians)", fontsize=11)
plt.ylabel("Amplitude", fontsize=11)
plt.axhline(0, color="black", linewidth=0.8, linestyle="-")
plt.grid(True, which="both", linestyle="--", alpha=0.5)

# Adding Legend
plt.legend(loc="upper right", frameon=True, shadow=True)

plt.tight_layout()
plt.show()
```
