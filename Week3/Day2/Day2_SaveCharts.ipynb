# Week 3 Day 2: Save Charts to File
Save generated charts as image files inside the project folder.

```python
import os
import matplotlib.pyplot as plt

# Create local folder right where this notebook lives
output_dir = "saved_charts"
os.makedirs(output_dir, exist_ok=True)

fig, ax = plt.subplots(figsize=(8, 4.5))

categories = ["Group A", "Group B", "Group C", "Group D"]
values = [45, 68, 82, 30]

ax.barh(categories, values, color="purple", alpha=0.8, edgecolor="black")
ax.set_title("Horizontal Bar Chart - Saved Output", fontweight="bold")
ax.set_xlabel("Score")
ax.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()

# Save output cleanly
output_path = os.path.join(output_dir, "Day2_HorizontalBarChart.png")
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.close()

print(f"Chart successfully saved to: {os.path.abspath(output_path)}")
```
