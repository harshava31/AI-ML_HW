# Week 3 Day 2: Scatter Plot
Plot two numerical columns in a scatter plot and observe the relationship between them.

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data: Study Hours vs. Test Scores
np.random.seed(42)
study_hours = np.random.uniform(1, 10, 50)
test_scores = study_hours * 8 + np.random.normal(0, 5, 50)

plt.figure(figsize=(8, 5))
plt.scatter(
    study_hours, test_scores, color="crimson", alpha=0.7, edgecolors="black"
)

plt.title("Study Hours vs. Test Score", fontsize=14, fontweight="bold")
plt.xlabel("Hours Studied", fontsize=12)
plt.ylabel("Test Score (%)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
