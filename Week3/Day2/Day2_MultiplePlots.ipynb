# Week 3 Day 2: Multiple Plot Practice
Create two or three separate charts and note the purpose of each chart.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [10, 15, 13, 18, 22]
expenses = [8, 11, 10, 12, 15]

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# Plot 1: Revenue Line Chart
axes[0].plot(months, revenue, color="green", marker="o", linewidth=2)
axes[0].set_title("Monthly Revenue ($K)", fontweight="bold")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Revenue")
axes[0].grid(True, linestyle=":", alpha=0.6)

# Plot 2: Expenses Bar Chart
axes[1].bar(months, expenses, color="coral", edgecolor="black", width=0.5)
axes[1].set_title("Monthly Expenses ($K)", fontweight="bold")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Expenses")
axes[1].grid(axis="y", linestyle=":", alpha=0.6)

plt.suptitle("Financial Performance Summary", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.show()
```
