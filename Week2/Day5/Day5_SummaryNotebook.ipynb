# Week 2 Data Science Fundamentals: NumPy & Pandas
This notebook summarizes key concepts covered in Week 2, including array manipulation, data cleaning, and dataset aggregation.

```python
import numpy as np
import pandas as pd

print("Libraries imported successfully!")
```

## 1. NumPy Array Creation & Operations

```python
# Array creation and basic statistics
arr = np.array([[10, 20, 30], [40, 50, 60]])

print(f"Shape: {arr.shape}")
print(f"Reshaped (3x2):\n{arr.reshape(3, 2)}")
print(f"Mean: {np.mean(arr):.2f}")
print(f"Standard Deviation: {np.std(arr):.2f}")
```

## 2. Pandas Data Cleaning & Aggregation

```python
# Sample sales dataset with missing data
data = {
    "Product": ["A", "B", "C", "A", "B"],
    "Price": [10, 15, 20, 10, 15],
    "Quantity": [5, np.nan, 3, 8, 2],
}
df = pd.DataFrame(data)

# Fill missing values and compute total revenue
df["Quantity"] = df["Quantity"].fillna(0)
df["Revenue"] = df["Price"] * df["Quantity"]

# Display summary dataframe and grouped total
display(df)
print("\nTotal Revenue per Product:")
display(df.groupby("Product")["Revenue"].sum().reset_index())
```
